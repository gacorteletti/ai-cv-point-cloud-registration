This script actually produces a Triangle Mesh instead of a Point Cloud. That is why we must use ```write_triangle_mesh``` instead of ```write_point_cloud``` (the latter would result in a TypeError).

Then, we need to convert this mesh to a cloud, which can be done by defining a function responsible to produce a cloud where its points are the vertices of the given mesh. We also copy colors and normals information.

OBS: this could be done using just the extra parameter **point_sample_factor** inside the ```draw_plotly``` method, which would define the percentage of points from the mesh we should sample to build our cloud. However, it's more consistent to follow our method and, if you want to sample the amount of points, perform a downsampling.

````python
def convert_mesh_to_pcd(input_mesh):

    mesh = copy.deepcopy(input_mesh)

    pcd = o3d.geometry.PointCloud()
    pcd.points = mesh.vertices

    if mesh.has_vertex_colors():
        pcd.colors = mesh.vertex_colors

    if mesh.has_vertex_normals():
        pcd.normals = mesh.vertex_normals

    return pcd
````

````python
output_mesh = o3d.io.read_triangle_mesh("output.ply")
output = convert_mesh_to_pcd(output_mesh)
print("Original #Points =", len(output.points))
````

Since the amount of points is too big to plot fully, we must perform a downsampling to be able to visualize it.

````python
output_down = output.voxel_down_sample(0.02)
print("Downsampled #Points =", len(output_down.points))
````

We can define here the same function from the previous Notebooks that allow us to visualize the relative pose between the clouds.

````python
def draw_registration_result(source, target, transformation, voxel_size=0.0):

    #create copies of both clouds to protect original data
    source_temp = copy.deepcopy(source)
    target_temp = copy.deepcopy(target)

    #downsample for faster visualization (voxel_size is in meters)
    if (voxel_size):
        source_temp = source_temp.voxel_down_sample(voxel_size)
        target_temp = target_temp.voxel_down_sample(voxel_size)

    #paint target cyan and source yellow
    source_temp.paint_uniform_color([1, 0.706, 0])
    target_temp.paint_uniform_color([0, 0.651, 0.929])

    #apply the transformation to the source cloud
    source_temp.transform(transformation)

    #plot target and transformed source clouds
    o3d.visualization.draw_plotly([source_temp, target_temp], width=900, height=600)
````

To observe the initial relative pose, we define the registration transformation as an identity matrix (so it does not affect the original cloud).

````python
draw_registration_result(source, target, np.eye(4), 0.02)
````

Now, in order to implement it, we need to <mark>initialize the model</mark> we selected and downloaded its weigths, which was **ResUNetBN2C** with:
* 1 input channel (only depth image, no RGB)
* 16 output channels (it obtains 16 features distributed over the image)
* Enable normalization of features
* Kernel dimension of 3x3x3 (size=3, D=3) for the first convolutional layer
    * OBS: the kernel size for the other layers are always fixed in 3 with the exception of the final layer and the one before that (the last deconvolutional layer — used for upsampling)

The model is set to evaluation, since we are not training it. 

````python
from model import load_model

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

model_name = 'ResUNetBN2C-32feat-3conv.pth'
model_path = os.path.join(fcgf_weights_folder, model_name)
checkpoint = torch.load(model_path, map_location=device)
config = checkpoint['config']

num_feats = 1
Model = load_model(config.model)
model = Model(
    num_feats,
    config.model_n_out,
    bn_momentum=0.05,
    normalize_feature=config.normalize_feature,
    conv1_kernel_size=config.conv1_kernel_size,
    D=3)
model.load_state_dict(checkpoint['state_dict'])
model.eval()

model = model.to(device)
````

Now we can use this model to obtain the features and, after that, we can color code the clouds based on the location of features in them. To do this, a function is defined following the example of the code provided by the demo.py script.

````python
#EXTRACT FEATURES
from util.visualization import get_colored_point_cloud_feature
from util.misc import extract_features

def ensure_dir(path):
  if not os.path.exists(path):
    os.makedirs(path, mode=0o755)


def extract_FCGF(cloud, model, voxel_size, device, target_path, save_fn):

    ensure_dir(target_path)

    pcd = copy.deepcopy(cloud)                                                                      # copy cloud to avoid modifying it
    xyz = np.array(pcd.points)                                                                      # extract coordinates of each point

    # Extract FCGF through a function given by the author
    xyz_down, features = extract_features(model, xyz, voxel_size=voxel_size, device=device, skip_check=True)

    # Obtain cloud with the features color coded
    vis_pcd = o3d.geometry.PointCloud()                                                             # initializes point cloud object
    vis_pcd.points = o3d.utility.Vector3dVector(xyz_down)                                           # defines points with coordinates from downsampled set
    vis_pcd = get_colored_point_cloud_feature(vis_pcd, features.detach().cpu().numpy(), voxel_size) # paint features with different colors for each type

    np.savez_compressed(os.path.join(target_path, save_fn),
                        points=np.array(pcd.points),
                        xyz=xyz_down,
                        features=features.detach().cpu().numpy())

    return xyz_down, features, vis_pcd
````

Once again, to visualize it, we must convert from TriangleMesh to PointCloud and perform the downsampling before plotting it. For this purpouse, the previous presented methods were combined into a function.

````python
def convert_mesh_to_pcd(input_mesh):

    mesh = copy.deepcopy(input_mesh)

    pcd = o3d.geometry.PointCloud()
    pcd.points = mesh.vertices

    if mesh.has_vertex_colors():
        pcd.colors = mesh.vertex_colors

    if mesh.has_vertex_normals():
        pcd.normals = mesh.vertex_normals

    return pcd


def plot_features(mesh, voxel_size=None):

    pcd = convert_mesh_to_pcd(mesh)
    print("Original #Points:", len(pcd.points))

    if voxel_size:
        pcd = pcd.voxel_down_sample(voxel_size)
        print("Downsampled #Points:", len(pcd.points))

    o3d.visualization.draw_plotly([pcd], width=900, height=600)
````

## 3.2 Correspondence Matching



````python
from util.pointcloud import compute_overlap_ratio, make_open3d_point_cloud, make_open3d_feature_from_numpy

def read_data(feature_path, name):
  data = np.load(os.path.join(feature_path, name + ".npz"))
  xyz = make_open3d_point_cloud(data['xyz'])
  feat = make_open3d_feature_from_numpy(data['features'])
  return data['points'], xyz, feat


def run_ransac(xyz0, xyz1, feat0, feat1, voxel_size):
  distance_threshold = voxel_size * 1.5
  result_ransac = o3d.pipelines.registration.registration_ransac_based_on_feature_matching(
      xyz0, xyz1, feat0, feat1, True, distance_threshold,
      o3d.pipelines.registration.TransformationEstimationPointToPoint(False), 4, [
          o3d.pipelines.registration.CorrespondenceCheckerBasedOnEdgeLength(0.9),
          o3d.pipelines.registration.CorrespondenceCheckerBasedOnDistance(distance_threshold)
      ], o3d.pipelines.registration.RANSACConvergenceCriteria(4000000, 0.999))
  return result_ransac.transformation


def do_single_pair_matching(feature_path, voxel_size):

    points_i, xyz_i, feat_i = read_data(feature_path, 'feat_source')
    points_j, xyz_j, feat_j = read_data(feature_path, 'feat_target')

    if len(xyz_i.points) < len(xyz_j.points):
        trans = run_ransac(xyz_i, xyz_j, feat_i, feat_j, voxel_size)
    else:
        trans = run_ransac(xyz_j, xyz_i, feat_j, feat_i, voxel_size)
        trans = np.linalg.inv(trans)
    #ratio = compute_overlap_ratio(xyz_i, xyz_j, trans, voxel_size)
    #logging.info("\t\t\tOverlap Ratio: %.4f" % (ratio))
    #if ratio > 0.3:
    return trans
    #else:
    #  return [False, i, j, s, np.identity(4)]
````
